This notebook verifies that the Python implementation of advanced features (Global MEM, Kinetics) aligns with the MATLAB playground logic.

```python
import matplotlib.pyplot as plt
import numpy as np
from chisurf.plugins.fcs.fcs_2d.kinetics import KineticsModel, calculate_population_evolution

print("Modules imported successfully.")
```

## 1. Verify Kinetic Modeling

We model a simple 2-state system: 1 <-> 2

```python
model = KineticsModel(n_states=2)
model.set_rate(from_state=0, to_state=1, rate=0.5)  # k12
model.set_rate(from_state=1, to_state=0, rate=0.1)  # k21

p_eq = model.get_equilibrium_populations()
print(f"Equilibrium populations: {p_eq}")

time_axis = np.linspace(0, 10, 100)
p_t = calculate_population_evolution(model.rate_matrix, np.array([1.0, 0.0]), time_axis)

plt.figure(figsize=(8, 4))
plt.plot(time_axis, p_t[0], label="State 1")
plt.plot(time_axis, p_t[1], label="State 2")
plt.xlabel("Time")
plt.ylabel("Population")
plt.legend()
plt.title("2-State Kinetic Evolution")
plt.show()
```
